```python
import json
with open('../dados/originais/data.json', 'r', encoding='utf-8') as f:
    arquivo = json.load(f)
    
    print(arquivo)
```

```python
# ajuste das construções e populações afetadas
# se o valor caiu, mesmo aumentando a altura da água, incorreto.
# assim, atribui-se o valor da medição anterior

construcoes_anteriores = {}
populacoes_anteriores = {}

for altura, bairros in arquivo.items():
    print(f"\nAltura: {altura}")

    for bairro, dados in bairros.items():
        construcoes_afetadas = dados['affected_buildings']
        populacao_afetada = dados['affected_population']

        print(f"\nBairro: {bairro}")
        print(f"Edificações afetadas: {construcoes_afetadas}")
        print(f"População afetada: {populacao_afetada}")

        c_anterior = construcoes_anteriores.get(bairro, 0)
        p_anterior = populacoes_anteriores.get(bairro, 0)

        if(construcoes_afetadas < c_anterior):
            construcoes_afetadas = c_anterior
        if(populacao_afetada < p_anterior):
            populacao_afetada = p_anterior

        # novo cálculo da porcentagem
        total_construcoes = dados['total_buildings']

        if total_construcoes > 0:
            porcentagem = (construcoes_afetadas / total_construcoes) * 100
        else:
            porcentagem = 0

        dados['affected_buildings'] = construcoes_afetadas
        dados['affected_population'] = populacao_afetada
        dados['percentage'] = round(porcentagem)

        construcoes_anteriores[bairro] = construcoes_afetadas
        populacoes_anteriores[bairro] = populacao_afetada


with open('../dados/processados/data_alterado.json', 'w', encoding='utf-8') as f:
    json.dump(arquivo, f, ensure_ascii=False, indent=4)
    print('Arquivo "data_alterado.json" agora possui as informações ajustadas')
```
